```python
# Question 1
#1: Prompt the user to enter the number of days
days = float(input("Enter the number of days: "))
#2: Define conversion constants
HOURS_PER_DAY = 24
MINUTES_PER_HOUR = 60
SECONDS_PER_MINUTE = 60
#3: Calculate total seconds (days * 24 * 60 * 60)
total_seconds = days * HOURS_PER_DAY * MINUTES_PER_HOUR * SECONDS_PER_MINUTE
#4: Display the result to the user
print(f"{days} day(s) is equal to {total_seconds:,.2f} seconds.")
```

```python
# Question 2
import math

#1: Prompt the user to enter the radius of the sphere
radius = float(input("Enter the radius of the sphere: "))

#2: Calculate the volume using the formula: V = (4/3) * π * r^3
# The '**' operator is used for exponentiation (r^3)
volume = (4 / 3) * math.pi * (radius ** 3)

#3: Output the calculated volume rounded to 4 decimal places
print(f"The volume of a sphere with radius {radius} is: {volume:.4f}")
```

```python
# Question 3
# Function to compute the area of a square
def calculate_area(side_length):
    return side_length ** 2

# Function to compute the perimeter of a square
def calculate_perimeter(side_length):
    return 4 * side_length

def main():
    #1: Ask the user to input the side length of the square
    side = float(input("Enter the side length of the square: "))

    #2: Ensure the side length is positive
    if side < 0:
        print("Side length cannot be negative.")
        return

    #3: Call the functions
    area = calculate_area(side)
    perimeter = calculate_perimeter(side)

    #4: Display the computed values
    print(f"\n--- Square Properties (Side = {side}) ---")
    print(f"Area: {area:}")
    print(f"Perimeter: {perimeter:}")

# Execute the main
if __name__ == "__main__":
    main()
```

```python
# Question 4
def check_character_case(char):
    """
    Evaluates the character and returns its case status.
    """
    if char.isupper():
        return "uppercase"
    elif char.islower():
        return "lowercase"
    elif char.isalpha():
        return "an alphabetic character of undetermined case"
    else:
        return "not an alphabetic letter"

def main():
    #1: Prompt user for a single character input
    user_input = input("Enter a single character: ")

    #2: Confirm that exactly one character was provided
    if len(user_input) != 1:
        print("Please enter exactly one character.")
        return

    #3: Call function and display classification
    result = check_character_case(user_input)
    print(f"The character '{user_input}' is {result}.")

if __name__ == "__main__":
    main()
```

```python
# Question 5
# The pseudocode was:
# BEGIN
# SET x TO 0, y TO 20
# REPEAT
#     SUBTRACT 4 FROM y
#     ADD 2/y TO x
# UNTIL y IS LESS THAN 6
# DISPLAY x
# END

#1: Initialize variables x and y
x = 0.0
y = 20

#2: Implement REPEAT ... UNTIL using a while True loop with a break condition
while True:
    y = y - 4
    x = x + (2 / y)

    if y < 6:
        break

#3: DISPLAY x
print(f"The final value of x is: {x}")
```

```python
#Question 6

#1: Initialize an empty list to store values
values = []
TOTAL_ITEMS = 5

#1: Use a loop to continually prompt for 5 numbers
print(f"Please enter {TOTAL_ITEMS} numbers:")
for i in range(1, TOTAL_ITEMS + 1):
    val = float(input(f"Enter value {i} of {TOTAL_ITEMS}: "))
    values.append(val)

#3: Calculate the sum and average
total_sum = sum(values)
average = total_sum / len(values)

#4: Display the populated array and calculated average
print("\n--- Results ---")
print(f"Populated array: {values}")
print(f"Average of the values: {average:}")
```
